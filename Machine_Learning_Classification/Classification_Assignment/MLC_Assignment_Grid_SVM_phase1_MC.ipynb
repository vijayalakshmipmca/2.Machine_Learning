```python
import numpy as np
import pandas as pd
import pickle as pc
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report, f1_score, roc_auc_score
from sklearn.preprocessing import StandardScaler
```

```python
Dataset = pd.read_csv("CKD.csv")
Dataset = pd.get_dummies(Dataset,dtype = int, drop_first = True)
Dataset
```

```python
Dataset['classification_yes'].value_counts()
```

```python
print("Dataset Columns : ", Dataset.columns)
```

```python
Independent = Dataset[['age', 'bp', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hrmo', 'pcv',
       'wc', 'rc', 'sg_b', 'sg_c', 'sg_d', 'sg_e', 'rbc_normal', 'pc_normal',
       'pcc_present', 'ba_present', 'htn_yes', 'dm_yes', 'cad_yes',
       'appet_yes', 'pe_yes', 'ane_yes']]
Dependent = Dataset['classification_yes']
print("Independent shape :" ,Independent.shape)
print("Dependent shape :" ,Dependent.shape)
```

```python
Dependent
```

```python
X_Train,X_Test,Y_Train,Y_Test = train_test_split(Independent,Dependent,test_size =0.30, random_state = 0)
```

```python
scx = StandardScaler()
X_Train = scx.fit_transform(X_Train)
X_Test = scx.transform(X_Test)
```

```python
grid_param = {'kernel' : ['poly','rbf','sigmoid','linear'] ,
              'C' : [0.1,1,100,500,1000,2000,3000] ,
              'gamma' : ['auto','scale']
             }
grid_classifier = GridSearchCV(SVC(probability = True), grid_param,refit = True, n_jobs = -1, scoring = 'f1_weighted')
grid_classifier.fit(X_Train,Y_Train)
```

```python
result = grid_classifier.cv_results_
#print(result)
grid_Y_Pred = grid_classifier.predict(X_Test)
Con_Mat = confusion_matrix(Y_Test, grid_Y_Pred)
print("Confusion Matrix : \n ", Con_Mat)
```

```python
fl_macro_avg = f1_score(Y_Test,grid_Y_Pred,average='weighted')
print("The f1 score-macro avg value for best parameter {}:".format(grid_classifier.best_params_),fl_macro_avg)
```

```python
clf = classification_report(Y_Test, grid_Y_Pred)
print("Classification Report : \n ", clf)
```

```python
roc_auc_score(Y_Test,grid_classifier.predict_proba(X_Test)[:,1])
```
