```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB, CategoricalNB, ComplementNB
from sklearn.metrics import confusion_matrix , classification_report
```

```python
Dataset = pd.read_csv("Social_Network_Ads.csv")
Dataset = pd.get_dummies(Dataset, dtype = int, drop_first = True)
```

```python
Dataset['Purchased'].value_counts()
```

```python
print("Before removal of column : ", Dataset.columns)
Dataset = Dataset.drop("User ID", axis=1)
print("After removal of column : ", Dataset.columns)
```

```python
Independent = Dataset[['Age', 'EstimatedSalary', 'Gender_Male']]
Dependent = Dataset ['Purchased']
Independent.shape
```

```python
Dependent
```

```python
X_Train, X_Test, Y_Train, Y_Test = train_test_split(Independent, Dependent, test_size = 0.30, random_state = 0)
```

```python
Classifier = GaussianNB()
Classifier.fit(X_Train,Y_Train)
Y_Pred = Classifier.predict(X_Test)
con_Metrics = confusion_matrix(Y_Test, Y_Pred)
print("Confusion_Matrix : \n", con_Metrics)
clf = classification_report(Y_Test, Y_Pred)
print("Classification_Report : \n", clf)
```

```python
Classifier = MultinomialNB()
Classifier.fit(X_Train,Y_Train)
Y_Pred = Classifier.predict(X_Test)
con_Metrics = confusion_matrix(Y_Test, Y_Pred)
print("Confusion_Matrix : \n", con_Metrics)
clf = classification_report(Y_Test, Y_Pred)
print("Classification_Report : \n", clf)
```

```python
Classifier = BernoulliNB()
Classifier.fit(X_Train,Y_Train)
Y_Pred = Classifier.predict(X_Test)
con_Metrics = confusion_matrix(Y_Test, Y_Pred)
print("Confusion_Matrix : \n", con_Metrics)
clf = classification_report(Y_Test, Y_Pred)
print("Classification_Report : \n", clf)
```

```python
Classifier = CategoricalNB()
Classifier.fit(X_Train,Y_Train)
Y_Pred = Classifier.predict(X_Test)
con_Metrics = confusion_matrix(Y_Test, Y_Pred)
print("Confusion_Matrix : \n", con_Metrics)
clf = classification_report(Y_Test, Y_Pred)
print("Classification_Report : \n", clf)
```

```python
Classifier = ComplementNB()
Classifier.fit(X_Train,Y_Train)
Y_Pred = Classifier.predict(X_Test)
con_Metrics = confusion_matrix(Y_Test, Y_Pred)
print("Confusion_Matrix : \n", con_Metrics)
clf = classification_report(Y_Test, Y_Pred)
print("Classification_Report : \n", clf)
```
