```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVR
from sklearn.metrics import r2_score
import pickle as pc
```

```python
Dataset = pd.read_csv ("insurance_pre.csv")
Dataset= pd.get_dummies(Dataset,dtype ="int", drop_first = True)
Dataset.columns
```

```python
Independent = Dataset [['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
Dependent = Dataset [['charges']]
```

```python
param_grid = {'kernel' : ['poly','rbf','sigmoid'] ,
              'C' : [0.1,1,100] ,
             # 'gamma' : ['auto','scale']
             }
             
regressor_grid =GridSearchCV(SVR(),param_grid,refit = True, verbose = 3, n_jobs = -1)
regressor_grid.fit(Independent, Dependent)
```

```python
Result = regressor_grid.cv_results_
print("The best parameter is {} :".format(regressor_grid.best_params_))
```

```python
table = pd.DataFrame.from_dict(Result)
table
```

```python
Age_Inp = float(input("Age:"))
BMI_Inp = float(input("BMI:"))
Children_Inp = float(input("Children:"))
Gender_Inp = float(input("Gender_Male_Or_Female(1 or 0):"))
Smoker_Inp = float(input("Smoker_Yes_Or_No (1 or 0):"))
PreInp = ([[Age_Inp,BMI_Inp,Children_Inp,Gender_Inp,Smoker_Inp]])
Future_Predict = regressor_grid.predict(PreInp)
print("Future Prediction : {}".format(Future_Predict))
```
