```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVR
from sklearn.metrics import r2_score
import pickle as pc
```

```python
Dataset = pd.read_csv ("insurance_pre.csv")
Dataset= pd.get_dummies(Dataset,dtype ="int", drop_first = True)
Dataset.columns
```

```python
Independent = Dataset [['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
Dependent = Dataset [['charges']]
X_Train, X_Test, Y_Train,Y_Test = train_test_split(Independent,Dependent,test_size = 0.3, random_state = 0)
```

```python
regressor =SVR(kernel = 'linear',C=1000,epsilon = 0.1)
regressor.fit(X_Train, Y_Train)
Y_Pred = regressor.predict(X_Test)
```

```python
r_value = r2_score(Y_Test, Y_Pred)
r_value
```

```python
Filename = "MLR_Assignment_SVM_Model.sav"
pc.dump(regressor,open(Filename,'wb'))
Loaded_file = pc.load(open ("MLR_Assignment_SVM_Model.sav",'rb'))
Result = Loaded_file.predict([[43,30.970,1,0,0]])
Result
```
