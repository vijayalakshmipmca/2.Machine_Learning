```python
import pandas as pd
from catboost import CatBoostRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import pickle as pc
```

```python
Dataset = pd.read_csv("50_Startups.csv")
#Dataset = pd.get_dummies(Dataset, dtype = "int" , drop_first = True)
Dataset.columns
```

```python
Independent = Dataset [['R&D Spend', 'Administration', 'Marketing Spend','State']]
Dependent = Dataset [['Profit']]
```

```python
cat_features = ['State']
```

```python
X_Train, X_Test, Y_Train, Y_Test = train_test_split(Independent,Dependent,test_size = 0.3, random_state = 0)
```

```python
regressor = CatBoostRegressor(iterations=500, learning_rate = .1,depth = 6,verbose=0)
regressor.fit(X_Train, Y_Train,cat_features=cat_features)
```

```python
Y_Pred = regressor.predict(X_Test)
Y_Pred
```

```python
r_value = r2_score(Y_Test, Y_Pred)
r_value
```

```python
Filename = "Machine_Learning_Regression_CatBoost_Model.sav"
Filename
```

```python
pc.dump(regressor,open(Filename,'wb'))
```
