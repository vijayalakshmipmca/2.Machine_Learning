```python
import pandas as pd
import joblib
```

```python
model = joblib.load("insurance_svr_pipeline.pkl")
sc_X = joblib.load('sc_X.pkl')
sc_Y = joblib.load('sc_Y.pkl')
```

```python
new_data = pd.DataFrame({
    'age':[43],
    'bmi':[30.970],
    'children':[1],
    'sex_male':[0],
    'smoker_yes':[0]
})
```

```python
new_data_scaled = sc_X.transform(new_data)
prediction = model.predict(new_data_scaled)

print(prediction)
```

```python
pred_original = sc_Y.inverse_transform(
    prediction.reshape(-1,1)
)

print(pred_original)
```
