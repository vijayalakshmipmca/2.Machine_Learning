```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
import pickle as pc
```

```python
Dataset = pd.read_csv("insurance_pre.csv")
Dataset = pd.get_dummies(Dataset, dtype = "int", drop_first = True)
Dataset.columns
```

```python
Independent = Dataset [['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
Dependent = Dataset [['charges']]
X_Train,X_Test,Y_Train,Y_Test = train_test_split (Independent,Dependent,test_size = 0.3, random_state = 0)
```

```python
scx = StandardScaler()
X_Train = scx.fit_transform(X_Train)
X_Test = scx.transform(X_Test)
```

```python
scy = StandardScaler()
Y_Train = scy.fit_transform(Y_Train)
Y_Test = scy.transform(Y_Test)
```

```python
param_grid = {'criterion' : ['squared_error', 'absolute_error','poisson'],
              'max_features' : ['sqrt','log2',None],
              'n_estimators' : [50,100,200,500,1000]}
regressor_grid = GridSearchCV(RandomForestRegressor(),param_grid, refit = True, verbose = 3, n_jobs = -1)
regressor_grid.fit(X_Train, Y_Train)
```

```python
Y_Pred = regressor_grid.predict(X_Test)
r_value = r2_score (Y_Test, Y_Pred)
r_value
```

```python
Result = regressor_grid.cv_results_
print("The best parameter is : {} ".format(regressor_grid.best_params_))
```

```python
Table = pd.DataFrame.from_dict(Result)
Table
```
